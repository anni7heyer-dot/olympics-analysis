```python
import pandas as pd
df = pd.read_csv("olympics.csv", sep=";")
df.shape
```

```python
df.head()
```

```python
check = df.groupby("ID")[["Height", "Weight"]].nunique()
(check > 1).sum()
#überprüfen ob das Gewicht oder die Größe sich geändert hat
```

```python
athletes = df[["ID", "Name","Sex", "Height", "Weight"]].drop_duplicates(subset="ID")
athletes = athletes.rename(columns={
    "ID":"athlete_id",
    "Name":"name",
    "Sex":"sex",
    "Height":"height",
    "Weight":"weight"
})

athletes.shape
athletes.head()
```

```python
athletes["athlete_id"].is_unique
```

```python
athletes.dtypes
```

```python
athletes.isna().sum()
```

```python
athletes.to_csv("athletes.csv", index=False)
#Athleten-Tabellen fertig
```

```python
df.head()
#jetzt Country
```

```python
country = df[["NOC","Country"]].drop_duplicates(subset="NOC")
country = country.rename(columns={
    "NOC":"noc",
    "Country":"country"
})

country.shape
country.head()
```

```python
country.dtypes
```

```python
country["noc"].is_unique
```

```python
country.to_csv("countries.csv", index=False)
#Country-Tabelle fertig
```

```python
df.head()
#Sport-Tabelle
```

```python
sport = df[["Sport"]].drop_duplicates().reset_index(drop=True)
sport["sport_id"] = sport.index + 1

sport = sport.rename(columns={"Sport": "sport_name"})
sport = sport[["sport_id", "sport_name"]]  # Reihenfolge wie im ERD

sport.head()
```

```python
sport["sport_id"].is_unique
sport.shape
```

```python
sport.to_csv("sports.csv", index=False)
```

```python
df.head()
#Event Tabelle
```

```python
events = df[["Event", "Sport"]].drop_duplicates().reset_index(drop=True)
events["event_id"] = events.index + 1

events = events.merge(sport, left_on="Sport", right_on="sport_name", how="left")
events = events.rename(columns={"Event": "event_name"})
events = events[["event_id", "event_name", "sport_id"]]

events.shape
events.head()
```

```python
events.to_csv("events.csv", index=False)
#event Tabelle done
```

```python
participations = df[["ID", "Event", "NOC", "Year", "Age", "Medal"]].copy()
participations = participations.reset_index(drop=True)
participations["participation_id"] = participations.index + 1
participations.head()
```

```python
participations = participations.merge(
    events, left_on="Event", right_on="event_name", how="left"
)
participations.head()
```
